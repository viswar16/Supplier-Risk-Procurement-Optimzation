### 2.3 Date Validation

Objective:
Validate the order, dispatch, and delivery dates
to ensure chronological consistency and support
accurate lead-time calculations.

```python
#create an o copy of the original data as it is preserved
df_validation = df.copy()
```

```python
#define date columns
date_columns = [
    'Order_Date',
    'Dispatch_Date',
    'Delivery_Date'
]
```

```python
for col in date_columns:
    df_validation[col] = pd.to_datetime(
    df_validation[col],
    errors = 'coerce'
    )
```

```python
#display the converted date and data type 
print(df_validation[date_columns].dtypes)
```

```python
df_validation[date_columns].head()
```

```python
#check invalid date columns 
invalid_date_counts = (
    df_validation[date_columns]
    .isna()
    .sum()
)
invalid_date_columns
```

```python
#check whether the dispatch has happened before the order 
dispatch_before_order = (
    df_validation['Dispatch_Date'] < df_validation['Order_Date']
)
dispatch_before_order.sum()
```

```python
#check whether the delivery has occured before dispatch
delivery_before_dispatch = (
    df_validation['Delivery_Date'] < df_validation['Delivery_Date']
)
delivery_before_dispatch.sum()
```

```python
#Calculate lead time components
#time from order placed to dispatch 
df_validation['processing_time_days'] = (
    df_validation['Dispatch_Date'] - df_validation['Order_Date']
).dt.days
df_validation['processing_time_days']
```

```python
#time from dispatch days to delivery dates
df_validation['transit_time_days'] = (
    df_validation['Delivery_Date'] - df_validation['Dispatch_Date']
).dt.days
df_validation['transit_time_days']
```

```python
# Total orders to delivery time 
df_validation['actual_lead_time_days'] = (
    df_validation['Delivery_Date'] - df_validation['Order_Date']
).dt.days
df_validation['actual_lead_time_days']
```

```python
#preview the new feature 
df_validation[
    ['Order_ID',
    'processing_time_days',
    'transit_time_days',
    'actual_lead_time_days']
    ].head()
```

```python
#validated the created lead time component
lead_time_mismatch = (
    df_validation['processing_time_days'] + df_validation['transit_time_days'] != df_validation['actual_lead_time_days']
)
lead_time_mismatch.sum()
```

### 2.4 Numerical Data Validation

Objective:
Validate numerical variables against business rules
to identify impossible or suspicious values before
supplier risk analysis.

```python
# check invalid quantites and order value 
invalid_quantity = (df_validation['Quantity_Ordered'] <= 0).sum()
invalid_quantity
```

```python
invalid_order_value = (df_validation['Order_Value_USD'] <=0).sum()
invalid_order_value
```

```python
# validate supplier reliability code 
invalid_reliability = (
    (df_validation['Supplier_Reliability_Score'] <0)
    |
    (df_validation['Supplier_Reliability_Score'] >1)
).sum()
print('Invalid_reliability_scores:', invalid_reliability)
```

```python
#check negative and missing delay values 
negative_delays = (
    (df_validation['Delay_Days'] < 0 )
).sum()
```

```python
missing_delays = (
    (df_validation['Delay_Days'].isnull())
).sum()
```

```python
print('Negative_delays:', negative_delays)
print('Missing_delays:',missing_delays)
```

```python
#check negative historical disruption count 
invalid_disruption_count = (
    df_validation['Historical_Disruption_Count'] < 0
).sum()
print('invalid disruption count:', invalid_disruption_count)
```

```python

# Consolidate numerical validation results

numerical_validation = pd.DataFrame({
    "Validation_Check": [
        "Non-positive quantities",
        "Non-positive order values",
        "Reliability outside 0-1",
        "Negative delay days",
        "Missing delay days",
        "Negative historical disruption counts"
    ],
    "Invalid_Record_Count": [
        invalid_quantity,
        invalid_order_value,
        invalid_reliability,
        negative_delays,
        missing_delays,
        invalid_disruption_count
    ]
})

display(numerical_validation)
```
